```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch
```

```python
model_name = "mansoorhamidzadeh/parsbert-persian-QA"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)
```

```python
context = "تهران پایتخت ایران است و جمعیت آن بیش از 18 میلیون نفر می‌باشد. برج میلاد در تهران قرار دارد."
question = "جمعیت تهران چقدر است؟"

inputs = tokenizer(question, context, return_tensors="pt")
with torch.no_grad():
    outputs = model(**inputs)

start = torch.argmax(outputs.start_logits).item()
end = torch.argmax(outputs.end_logits).item()

answer = tokenizer.decode(
    inputs["input_ids"][0][start:end + 1],
    skip_special_tokens=True
)

print(answer)
```
